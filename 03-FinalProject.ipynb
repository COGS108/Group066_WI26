### Descriptions of any dataset shortcomings relevant to the project

This dataset has several limitations that may affect statistical analysis. First, installs are recorded as milestone categories (e.g., “1M+” or “10M+”) rather than exact counts, which reduces precision when converting them into numeric values for modeling. Second, the dataset is cross-sectional and represents only a single snapshot in time, so it cannot capture dynamic changes in rank, installs, or growth, nor establish causal relationships.  

There is also potential multicollinearity between installs and total ratings, since games with more downloads tend to accumulate more ratings. This may complicate regression interpretation and affect the relative importance of predictors. Finally, the dataset does not include other important ranking factors, such as revenue or user retention, which could result in omitted variable bias. Despite these limitations, the dataset is suitable for exploring predictive features of rank within the genre-specific Top 100 charts, which aligns with the research question.


### Top Games on Google Play Store 
This dataset contains the top 100 ranked Android games from each of 17 Google Play Store game categories (Action, Puzzle, Casino, etc.), totaling 1,730 rows before cleaning.

**Total ratings** is an integer count of how many users have left a star rating for the game. In this dataset it ranges from ~33,000 to ~86 million (Garena Free Fire). The median is about 430,000. To put that in context, a game with 100K ratings is reasonably well-known; a game with 1M+ ratings is a mainstream hit; and anything above 10M is a global phenomenon (e.g. PUBG, Free Fire, Subway Surfers). This metric is the best proxy we have for a game's overall popularity and user engagement, since actual download counts are bucketed.

**Installs** are reported as coarse string buckets — "100.0 k", "1.0 M", "10.0 M", "100.0 M", etc. — with only 9 distinct levels across the whole dataset. Nearly half of all games (47%) sit in the 10.0 M bucket, so this variable has extremely low resolution (i.e. a game listed as "10.0 M" installs could have anywhere from 10 million to 49.9 million actual installs).|

**Average rating** is reported as an integer — the only values present are 2, 3, and 4. This is the Play Store's displayed rating rounded to the nearest whole number, which destroys almost all signals (91% of games show a 4, 9% show a 3, and exactly one game shows a 2). This column is essentially useless for any comparative analysis.

**Growth (30 days) and growth (60 days)** represent the percentage increase in total ratings over the last 30 and 60 days, respectively. Typical values are small: the median 30-day growth is just 0.5%, reflecting the steady trickle of ratings that established games receive. Values of 5–10% suggest a game gaining meaningful momentum (perhaps from a marketing push). However, a handful of entries show absurd values—one game reports 227,106% 30-day growth—which almost certainly means it went from near-zero ratings to some modest number, making the percentage astronomically large despite a small absolute change.

**Star rating breakdowns (5-star through 1-star counts)** give the raw count of ratings at each level. These are highly correlated with total ratings (r > 0.93 for all five), since bigger games simply get more of everything. What's more analytically interesting is the ratio — for instance, a game where 1-star ratings make up 15%+ of its total (like Garena Free Fire at ~14.5%) signals polarizing quality or aggressive monetization that frustrates a vocal minority, even if the weighted average looks fine.

**The price** is in US dollars. The overwhelming majority of games (99.6%) are free; only 7 paid games appear in the dataset, ranging from \\$0.99 to \$7.49. This reflects the reality of the mobile gaming market: the top charts are dominated by free-to-play titles that monetize through ads and in-app purchases rather than upfront cost.

A key consideration for this dataset is that it includes only games ranked within the Google Play Store’s genre-specific Top 100 charts, which introduces selection bias by excluding lower-ranked or newly released games. As a result, the analysis focuses on explaining relative rank among already successful titles rather than general game success. Because the dataset contains only high-performing games, several quantitative features relevant to the research question, such as average rating and install milestones, exhibit limited variation, which may reduce their predictive power in statistical inference. Additionally, while rank values are expected to be unique within each genre, the dataset contains repeated rank numbers (e.g., multiple entries with the same rank such as 50), creating ambiguity in how rank is assigned and interpreted. Without additional information, it is unclear whether these repeated ranks correspond to ties, separate genre-specific rankings, or another ranking mechanism, which may affect the precision of rank-based modeling. Finally, the Play Store ranking system may incorporate factors not captured in the dataset, such as proprietary algorithms or platform-driven promotion, meaning rank may not be determined solely by observable metrics like total ratings, installs, or growth rates. Despite these limitations, the dataset remains appropriate for analyzing the relative importance of quantitative features in predicting rank within genre-specific Top 100 charts, which is the focus of this study.

#### Clean the Data

```python
#dropping all duplicated games in the dataframe
df_clean = df[~df.duplicated(subset=['category', 'rank'], keep=False)].copy()
df_clean.loc[:, 'rank_score'] = 101 - df_clean['rank']
```

##### We have unstandardized install counts, which we will convert from e.g. 10.0 M -> 10000000

```python
def convert_installs_to_value(value):
    if pd.isna(value):
        return value
    
    value = str(value).strip()
    
    if value.endswith('M') or value.endswith('m'):
        return float(value[:-1])*1000000
    elif value.endswith('k') or value.endswith('K'):
        return float(value[:-1])*1000
    else:
        # If no suffix, assume it's already a number
        return float(value)
```

##### We have some outliers within each category, which we will remove before doing a log transform

```python
# Apply to your column
df_clean.loc[:, 'installs'] = df_clean['installs'].apply(convert_installs_to_value)

log_transform_cols = ['total ratings', 'growth (30 days)', 'growth (60 days)',
                      '5 star ratings', '4 star ratings', '3 star ratings',
                      '2 star ratings', '1 star ratings']

def remove_iqr_outliers_by_category(data, group_col, columns, whisker_width=1.5):
    keep_mask = pd.Series(True, index=data.index)
    summary_rows = []

    for category_value, group in data.groupby(group_col):
        group_keep_mask = pd.Series(True, index=group.index)

        for col in columns:
            q1 = group[col].quantile(0.25)
            q3 = group[col].quantile(0.75)
            iqr = q3 - q1

            if pd.isna(iqr) or iqr == 0:
                continue

            lower = q1 - whisker_width * iqr
            upper = q3 + whisker_width * iqr
            group_keep_mask &= group[col].between(lower, upper, inclusive='both')

        removed_count = int((~group_keep_mask).sum())
        summary_rows.append({
            'category': category_value,
            'rows_removed': removed_count,
            'rows_remaining': int(group_keep_mask.sum())
        })
        keep_mask.loc[group.index] = group_keep_mask

    summary = pd.DataFrame(summary_rows).sort_values(['rows_removed', 'category'], ascending=[False, True])
    return data.loc[keep_mask].copy(), summary

rows_before_outlier_filter = len(df_clean)
df_clean, category_outlier_summary = remove_iqr_outliers_by_category(
    df_clean,
    group_col='category',
    columns=log_transform_cols
)

# category_outlier_summary[category_outlier_summary['rows_removed'] > 0]
```

```python
for col in log_transform_cols:
    df_clean.loc[:, f'{col}_log'] = np.log(df_clean[col])
```

```python
# Standardize numeric columns within each genre so values are comparable across categories.
df_standardized = df_clean.copy()
numeric_cols = df_standardized.select_dtypes(include=[np.number]).columns.tolist()
cols_to_standardize = [col for col in numeric_cols if col not in ['rank', 'rank_score']]


def zscore_within_genre(series):
    std = series.std(ddof=0)
    if pd.isna(std) or std == 0:
        return pd.Series(0.0, index=series.index)
    return (series - series.mean()) / std


df_standardized[cols_to_standardize] = df_standardized.groupby('category')[cols_to_standardize].transform(zscore_within_genre)

# Save the cleaned + standardized dataset
df_standardized.to_csv("data/02-processed/android-games-processed.csv", index=False)
df_standardized.head()
```

```python
# Reverse rank so higher = better
df_standardized['rank_reversed'] = df_standardized['rank'].max() + 1 - df_standardized['rank']

corr_cols = ['rank_reversed', 'total ratings_log', 'installs', 'average rating',
             'growth (30 days)', 'growth (60 days)', 
             '5 star ratings_log', '4 star ratings_log', 
             '3 star ratings_log', '2 star ratings_log', '1 star ratings_log']

corr_matrix = df_standardized[corr_cols].corr(method='spearman')

fig, ax = plt.subplots(figsize=(6, 5))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

sns.heatmap(
    corr_matrix,
    mask=mask,
    annot=True,
    fmt='.2f',
    annot_kws={'size': 6},
    cmap='RdBu_r',
    center=0,
    square=True,
    linewidths=0.5,
    ax=ax
)
ax.tick_params(labelsize=7)
ax.set_title('Spearman Correlation Heatmap (reversed rank, so worst-best is 1-100)', fontsize=9)
plt.tight_layout()
plt.show()

```

The overall Spearman heatmap shows a clear popularity pattern. `rank_reversed` is almost perfectly correlated with `total ratings_log` (0.994), strongly correlated with `5 star ratings_log` (0.963), and still strongly positively correlated with the lower-star count columns as well. This means that, across all categories combined, better-ranked games are primarily the games with the largest rating volume. `installs` is only moderately related to `rank_reversed` (0.328), while `average rating` and both growth variables are close to zero, so overall chart position appears to reflect scale much more than rating quality or short-term growth.

The rest of the matrix reinforces this interpretation. `total ratings_log` and the star-count columns are all highly correlated with one another, which shows that these variables mostly move together as different measures of overall game size. `growth (30 days)` and `growth (60 days)` are also nearly identical (0.987), so they are capturing the same signal. Taken together, the overall heatmap suggests that the strongest general relationship is: games rank better when they attract more engagement, not necessarily when they have slightly higher average ratings.


```python
sns.scatterplot(data=df_standardized, x='rank_reversed', y='total ratings_log')
plt.title('Rank Reversed vs Total Ratings Log')
plt.show()
```

This scatterplot makes the main result from the overall heatmap easier to see visually: as `rank_reversed` increases, `total ratings_log` rises almost monotonically. In other words, games with better ranks tend to have substantially more total ratings, and the upward curve is especially clear among the highest-ranked games. The tight pattern also explains why the overall Spearman correlation between these two variables is so high.

The plot suggests that total rating volume is one of the strongest observable markers of chart success in this dataset. At the same time, the curved shape shows that the relationship is not perfectly linear, so the scatterplot is most useful as evidence of a strong positive trend rather than a constant one-to-one increase. This visual supports the broader conclusion that rank is more closely tied to engagement scale than to small differences in average rating or short-term growth.


#### Section 2 of EDA - Exploring Feature Relationships Across Genres

Because the overall heatmap combines many different game categories, it can hide patterns that only appear within specific genres. This section breaks the data down by `category` and computes separate correlation heatmaps for each genre so we can compare whether the same relationships hold within categories. The goal is to see which patterns are consistent across genres and which ones vary depending on the type of game.


```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Make sure 'rank_reversed' exists in df_standardized
df_standardized['rank_reversed'] = df_standardized['rank'].max() + 1 - df_standardized['rank']

# Columns to include in correlation
corr_cols = ['rank_reversed', 'total ratings_log', 'installs', 'average rating',
             'growth (30 days)', 'growth (60 days)',
             '5 star ratings_log', '4 star ratings_log',
             '3 star ratings_log', '2 star ratings_log', '1 star ratings_log']

categories = df_standardized['category'].unique()
n_cats = len(categories)
n_cols = 2
n_rows = (n_cats + 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(n_cols * 6, n_rows * 5))
axes = axes.flatten()

for i, category_value in enumerate(categories):
    subset = df_standardized[df_standardized['category'] == category_value]
    corr_matrix = subset[corr_cols].corr(method='spearman')
    mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

    sns.heatmap(
        corr_matrix,
        mask=mask,
        annot=True,
        fmt='.2f',
        annot_kws={'size': 6},
        cmap='RdBu_r',
        center=0,
        square=True,
        linewidths=0.5,
        ax=axes[i]
    )
    axes[i].tick_params(labelsize=7)
    axes[i].set_title(f'Category: {category_value}', fontsize=9)

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Spearman Correlation Heatmaps by Category', fontsize=11, y=1.01)
plt.tight_layout()
plt.show()
```

The category-specific heatmaps reveal an even stronger within-genre pattern than the overall matrix. In essentially every category shown, `rank_reversed` and `total ratings_log` are perfectly or nearly perfectly aligned, meaning the rank ordering inside each category is almost the same as the ordering by total ratings. The same is true, to a slightly weaker degree, for the star-count columns: within categories, top-ranked games usually have the most 5-star ratings, and they also tend to have more 4-, 3-, 2-, and even 1-star ratings simply because they have much larger audiences overall.

There are also useful differences across categories. The relationship between `installs` and rank is much stronger in categories such as GAME CASINO, GAME SPORTS, GAME TRIVIA, GAME WORD, and GAME RACING, but weaker in categories such as GAME ACTION, GAME ADVENTURE, GAME EDUCATIONAL, GAME PUZZLE, GAME ROLE PLAYING, and GAME SIMULATION. `average rating` stays weak in most genres and is often negatively related to the lower-star count columns, especially in GAME TRIVIA, GAME MUSIC, GAME EDUCATIONAL, GAME ROLE PLAYING, and GAME SIMULATION, which suggests that review quality is somewhat separate from raw popularity. Finally, growth is the least consistent genre-level signal: `growth (30 days)` and `growth (60 days)` are always very strongly correlated with each other, but their relationship with rank is generally small and sometimes slightly negative. Overall, the within-category takeaway is that rank is driven most strongly by engagement volume, while the main differences across genres are how closely installs track rank and how strongly average rating separates from low-star review counts.
